```python
import torch

print(torch.__version__)
```

```python
# Creating Tensors

x = torch.tensor([1,2,3])

print(x)
print("Shape",x.shape)
print("Dtype", x.dtype)
print("Device", x.device)
```

```python
# Creating tensors with different Values

random_tensor = torch.rand(3, 4)
zeros_tensor = torch.zeros(2, 3)
ones_tensor = torch.ones(2, 3)

print("Random tensor:\n", random_tensor)
print("Shape:", random_tensor.shape)

print("\nZeros tensor:\n", zeros_tensor)

print("\nOnes tensor:\n", ones_tensor)
```

```python
x_data = torch.tensor([[1, 2], [3, 4]])

ones_like_x = torch.ones_like(x_data)
random_like_x = torch.rand_like(x_data, dtype=torch.float)

print("x_data:\n", x_data)
print("ones_like:\n", ones_like_x)
print("rand_like:\n", random_like_x)
```

```python
# concatenation
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])

print("dim=0:")
print(torch.cat([a, b], dim=0))

print("\ndim=1:")
print(torch.cat([a, b], dim=1))
```

```python
# matrix multiplication
A = torch.tensor([[1, 2], [3, 4]])
B = torch.tensor([[5, 6], [7, 8]])

C = A @ B

print(C)
print("Shape:", C.shape)
```

```python
import torch

x = torch.tensor(3.0, requires_grad=True)
u = x ** 2
v = u * 5
v.backward()

print("x =", x)
print("u =", u)
print("v =", v)
print("gradient =", x.grad)
```

```python
import torch

x = torch.tensor(3.0, requires_grad=False )
u = x ** 2
v = u * 5
v.backward()

print("x =", x)
print("u =", u)
print("v =", v)
print("gradient =", x.grad)
```
